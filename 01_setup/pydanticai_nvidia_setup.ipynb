# PydanticAI and Nvidia Build models

PydanticAI is a framework to create applications with AI models. PydanticAI is model agnostic and can use several models. 
In this lecture we will use Nvidia Builds models as they provide good quota limits (40 requests per minute as of this writing). 

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os

# chat completion style same as that of OpenAI, which Nvidia models support
model = OpenAIChatModel(
    "nvidia/nemotron-3.5-lightning-30b-a3b",  # any model from build.nvidia.com
    provider=OpenAIProvider( # connection details to nvidia
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"],  
    ),
)

# wraps a model
agent = Agent(
    model,
)
result = await agent.run("What is the capital of Sweden?")
print(result.output)
```
